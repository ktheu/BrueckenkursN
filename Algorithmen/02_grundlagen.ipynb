### while-Schleife

Der Rumpf der while-Schleife wird solange durchlaufen, bis die while-Bedingung sich zu False auswertet. 

```python
x = 10
while x >= 4:
    print(x)
    x -= 2
print("End")
```

### for-Schleife mit range()

Die Funktion range() liefert Aufzählungen von ganzen Zahlen und kann ein, zwei oder drei Parameter erhalten.

```python
# der Bereich beginnt bei 0, endet bei 5 (5 ist ausgeschlossen)
for k in range(5):
    print(k)
```

```python
# der Bereich beginnt bei 2, endet bei 7 (7 ist ausgeschlossen)
for k in range(2,5):
    print(k)
```

```python
# der Bereich beginnt bei 3, endet bei 9 (9 ist ausgeschlossen) mit Schrittweite 2
for i in range(3, 9, 2):
    print(i)
```

```python
# der Bereich beginnt bei 5, endet bei -1 (-1 ist ausgeschlossen) mit Schrittweite -1
for i in range(5, -1, -1):
    print(i)
```

```python
# Berechne die Summe aller ungeraden Zahlen von 1 bis 100:
summe = 0
for i in range(1, 100, 2):
    summe += i
print(f"Die Summe aller ungeraden Zahlen von 1 bis 100 ist: {summe}")
```

**Faustregel:** For-Schleifen nutzen wir meist dann, wenn die Anzahl der Schleifendurchgänge bekannt ist, andernfalls nutzen wir while-Schleifen.

### break und continue

Mit *break* verlassen wir eine Schleife und setzen den Ablauf hinter der Schleife fort. Mit *continue* hören mit dem laufenden Schleifendurchgang auf und beginnen den nächsten Schleifendurchgang. *break* und *continue* können nur in Schleifen verwendete werden.

```python
x = 2
while True:
    print(x)
    if x == 4: 
        break
    x+=1
print('End')
```

```python
for i in range(8):
    if i % 2 == 0: 
        continue
    print(i)
    
```

```python
# Die Schleife läuft, bis der Benutzer 0 eingibt
print('Gib nacheinander Zahlen ein (0 für Ende)')
summe = 0
while True:
    zahl = int(input())
    if zahl == 0:
        break
    summe = summe + zahl
print(f'Die Summe aller eingegebenen Zahlen ist {summe}')
     
```

### Schleifen für Strings und Listen

Zwei Arten, die Zeichen eines Strings oder die Elemente einer Liste zu durchlaufen:

##### Laufvariable über den Index

```python
# Die Laufvariable i geht über die Indizes des Strings
s = 'Habicht'
for i in range(len(s)):
    print(s[i])
```

```python
# Die Laufvariable i geht über die Indizes der Liste
a = [3, 4, 7, 2, 8]
for i in range(len(a)):
    print(a[i])
```

##### Laufvariable über die Zeichen/Elemente

```python
# Die Laufvariable c geht über die Zeichen des Strings
s = 'Habicht'
for c in s:
    print(c)
```

```python
# Die Laufvariable k geht über die Elemente der Liste
a = [3, 4, 7, 2, 8]
for k in a:
    print(k)
```

### Zufallszahlen

Das Modul random erzeugt Pseudozufallszahlen. Sie sehen zufällig aus, aber nach $2^{19937}-1$ Zahlen wiederholen sie sich.

```python
import random
for i in range(5):
    x = random.random()       # 0 <= x < 1
    print(x)
```

Mit der Funktion **seed** lässt sich der Zufallsgenerator initialisieren. Er erzeugt dann, je nach seed, immer die gleiche Zahlenfolge.

```python
random.seed(12)
for i in range(5):
    x = random.random()       # 0 <= x < 1
    print(x)
```

```python
random.randint(0,10)         #  ganze Zahl n: 0 <= n <= 10 
```

```python
random.random()              #  Dezimalzahl x:  0 <= x < 1
```

```python
random.uniform(2,10)         #  Dezimalzahl x: 2 <= x <= 10
```

### Funktionen

In Funktionen fassen wir Codeabschnitte zu einer Einheit zusammen.

<img src='./bilder/funktionen.png' width='401'>  

```python
def calc(x, y):
    z = x + y
    z = 3*z + 1
    return z
```

```python
a = calc(3,5)
print(a)
```

In einem *docstring*  können wir beschreiben, wie die Funktion aufgerufen wird und was sie zurückgibt.

```python
def div(x, y):
    '''
    x, y: zwei positive Zahlen
    returns: die kleinere dividiert durch die größere Zahl

    >>> div(5, 10)
    0.5

    >>> div(8, 2)
    0.25

    >>> div(4, 4)
    1.0
    '''
    if x < y:
        return x/y
    else:
        return y/x
```

### Dateien lesen

Vor dem Lesen oder Schreiben in eine Datei öffnen wir die Datei mit *open*. Nach der Verarbeitung schließen wir die Datei mit *close*.

Mit **readline** lesen wir eine Zeile. Der Zeilenvorschub wird mit übernommen. Mit *strip()* entfernen wir den Zeilenvorschub. 
Mit dem nächsten *readline* wird die nächste Zeile gelesen. Die Funktion *int* verwandelt die Eingabe in eine ganze Zahl und entfernt den Zeilenvorschub.

```python
%%writefile input1.txt
Hallo
5
```

```python
'''
Inhalt der Datei input1.txt:

Hallo
5
'''
f = open('input1.txt')    
s = f.readline().strip()           # 
x = int(f.readline())
f.close()
print(s)
print(x)
```

```python
%%writefile input2.txt
Daß üöä ÖÄÜ
```

Wenn wir bei Umlauten oder ß seltsame Zeichen sehen, hilft meist beim *open* der Parameter *encoding='utf-8'*.

```python
'''
Inhalt der Datei input2.txt:

Daß üöä ÖÄÜ
'''
f = open('input2.txt',encoding='utf-8')    
s = f.readline().strip()         
f.close()
print(s)
```

```python
'''
Die Datei output.txt bekommt folgenden Inhalt:
Die Summe von 5 und 10 ist 15
Das Produkt von 5 und 10 ist 50
'''
f = open('output.txt',encoding='utf-8',mode='w')
x = 5
y = 10
print(f'Die Summe von {x} und {y} ist {x+y}.', file=f)
print(f'Das Produkt von {x} und {y} ist {x*y}.', file=f)
f.close()
```
